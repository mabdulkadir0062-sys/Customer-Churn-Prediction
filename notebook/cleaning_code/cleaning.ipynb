```python
import pandas as pd
from pathlib import Path


FILE_PATH = Path("customer_churn.csv")
df = pd.read_csv(FILE_PATH)

print(
    df.head(20)
)

print("==== Duplicates ====")

print(
    df.duplicated().sum()
)

print("\n==== NULL Values ====")

print(
    df.isnull().sum()
)

print("\n==== Column types ====")

print(
    df.dtypes
)

print("\n==== Describe ====")

print(
    df.describe()
)

df = df.drop(columns=["customer_id"])

print(
    df.head(20)
)

df.to_csv(
    "customer_churn_clean.csv",
    index=False
)

```
